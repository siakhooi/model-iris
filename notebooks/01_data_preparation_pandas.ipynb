## 4. Transform Data

### 4.1 Convert fields type

```python
# Convert species column to uppercase string
df['species'] = df['species'].astype(str).str.upper()
```

### 4.2 Encoding

```python
from sklearn.preprocessing import LabelEncoder
import joblib

le = LabelEncoder()
df['species_label'] = le.fit_transform(df['species'])

print(df[['species', 'species_label']].head())
df = df.drop(columns=['species'])
df
```

## 6. Save Prepared Data

### prepare output directory

```python
output_dir = PANDA_PREPARED_DIR

import os
os.makedirs(output_dir, exist_ok=True)
```

### Save Label Encoding

```python
label_encoding_path = PANDA_LABEL_ENCODING_PATH

# Save encoder for later use
joblib.dump(le, label_encoding_path)
```

```python
label_encoding_path = PANDA_LABEL_ENCODING_PATH

# read pkl file and print content
le = joblib.load(label_encoding_path)
print(le.classes_)

label_mapping = {i: label for i, label in enumerate(le.classes_)}
print(label_mapping)
```

### Save as CSV

```python
csv_path = PANDA_PREPARED_DATA_PATH

df.to_csv(csv_path, index=False)
print(f"✓ Saved CSV: {csv_path}")
```

### Save as Parquet

```python
parquet_path = PANDA_PREPARED_DATA_PARQUET_PATH

df.to_parquet(parquet_path, index=False)
print(f"✓ Saved Parquet: {parquet_path}")
```

### Verify saved files

```python
print("Saved files:")
for f in os.listdir(output_dir):
    filepath = os.path.join(output_dir, f)
    size = os.path.getsize(filepath)
    print(f"  - {f} ({size} bytes)")
```
