# Model Training with scikit-learn

This notebook:
1. Loads prepared data from parquet
2. Trains a neural network model using scikit-learn/LogisticRegression
3. Exports the model to pickle format

## Import Libraries

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
import numpy as np
import os
```

# Constants

```python
INPUT_PATH = "../target/prepared/prepared_data.parquet"
FEATURE_COLS = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
MODELS_DIR = "../target/models"
SCALER_JOBLIB_FILENAME="scaler_sklearn.joblib"
MODEL_JOBLIB_FILENAME="model_sklearn.joblib"
MODEL_PICKLE_FILENAME="model_sklearn.pkl"
```

## Load Data from Parquet

```python
input_path = INPUT_PATH
df = pd.read_parquet(input_path)
```

```python
print(f"Data shape: {df.shape}")
```

```python
df.head()
```

```python
df.info()
```

## Prepare Features and Labels

```python
# Separate features (X) and target (y)
X = df[FEATURE_COLS].values.astype(np.float32)
y = df['species_label'].values.astype(np.int64)

print(f"Features shape: {X.shape}")
print(f"Labels shape: {y.shape}")
print(f"\nFeatures:\n{X}")
print(f"\nLabels: {y}")
```

# standardize the features

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled=scaler.fit_transform(X)
```

## Build the Neural Network Model

```python
model = LogisticRegression()
```

## Train the Model

```python
model.fit(X_scaled, y)
```

## Save Model with joblib and pickle

### Prepare Models directory

```python
models_dir = MODELS_DIR
os.makedirs(models_dir, exist_ok=True)
```

### Save Scaler

```python
import joblib

models_dir = MODELS_DIR
scaler_joblib_filename=SCALER_JOBLIB_FILENAME

scaler_path = os.path.join(models_dir, scaler_joblib_filename)
joblib.dump(scaler, scaler_path)
print(f"scikit-learn scaler saved to: {scaler_path}")
```

### Save model using joblib

```python
import joblib

models_dir = MODELS_DIR
model_joblib_filename=MODEL_JOBLIB_FILENAME

joblib_path = os.path.join(models_dir, model_joblib_filename)
joblib.dump(model, joblib_path)
print(f"scikit-learn model saved to: {joblib_path}")
```

### Save model using pickle

```python
import pickle

models_dir = MODELS_DIR
model_pickle_filename=MODEL_PICKLE_FILENAME

pickle_path = os.path.join(models_dir, model_pickle_filename)
with open(pickle_path, "wb") as f:
    pickle.dump(model, f)
print(f"scikit-learn model (pickle) saved to: {pickle_path}")
```
